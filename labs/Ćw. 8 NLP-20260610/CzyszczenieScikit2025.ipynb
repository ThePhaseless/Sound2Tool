#Przygotowanie danych tekstowych za pomocą scikit-learn

```python
from sklearn.feature_extraction.text import CountVectorizer
# lista dokumntów tekstowych
text = ["The quick brown fox jumped over the lazy dog."]
# transformacja
vectorizer = CountVectorizer()
# tokenizacja i budowa słownika
vectorizer.fit(text)
# podsumowanie
print(vectorizer.vocabulary_)
# zakodowany wektor
vector = vectorizer.transform(text)
# podsumowanie
print(vector.shape)
print(type(vector))
print(vector.toarray())
```

Widzimy, że w słownictwie jest 8 słów, dlatego zakodowane wektory mają  długość 8. Możemy zobaczyć, że zakodowany wektor jest macierzą rzadką. Na koniec możemy  zobaczyć tablicową wersję zakodowanego wektora pokazującą liczbę 1 wystąpienia dla każdego słowa z  wyjątkiem the (indeks 7), które ma wystąpienie 2.

```python
# zakodowanie dokumentu ze słowem ze słownika i słowem spoza słownika. Słowo, które nie jest w słowniku, zostaje zignorowane.
text2 = ["the puppy"]
vector = vectorizer.transform(text2)
print(vector.toarray())
```

TfidfVectorizer TFIDF-term frequency (podsumowuje, jak często dane słowo pojawia się w słowniku), inverse document frequency-zmniejsza skalę słów, które często pojawiają się w dokumentach

```python
from sklearn.feature_extraction.text import TfidfVectorizer
# list of text documents
text = ["The quick brown fox jumped over the lazy dog.",
"The dog.",
"The fox"]
# create the transform
vectorizer = TfidfVectorizer()
# tokenize and build vocab
vectorizer.fit(text)
# summarize
print(vectorizer.vocabulary_)
print(vectorizer.idf_)
# encode document
vector = vectorizer.transform([text[0]])
# summarize encoded vector
print(vector.shape)
print(vector.toarray())
```

Powyższe wyniki/wartości są znormalizowane do zakresu 0 do 1. Reprezentacja nadaje się bezpośrednio do większości algorytmów

```python
#Haszowanie
from sklearn.feature_extraction.text import HashingVectorizer
# Lista dokumntów tekstowych
text = ["The quick brown fox jumped over the lazy dog."]
# create the transform (zmień n_features na 4, 10)
vectorizer = HashingVectorizer(n_features=20)
# encode document
vector = vectorizer.transform(text)
# summarize encoded vector
print(vector.shape)
print(vector.toarray())

import pandas as pd
display(pd.DataFrame(vector.toarray()))
```

Używamy CountVectorizer dla 3 dokumentów.

```python
from sklearn.feature_extraction.text import CountVectorizer
# Lista dokumntów tekstowych
text = ["The quick brown fox jumped over the lazy dog.",
"The dog.",
"The fox"]
# create the transform
vectorizer = CountVectorizer()
# tokanizacja i budowa słownictwa
vectorizer.fit(text)
# summarize
document_term_matrix = vectorizer.transform(text)

# Podsumowanie
print("Kształt macierzy dokument-tokeny:", document_term_matrix.shape)
print("Tablica:")
print(document_term_matrix.toarray())

```
