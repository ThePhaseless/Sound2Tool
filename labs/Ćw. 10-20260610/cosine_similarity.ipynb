```python
import torch
# Definiujemy 4 tensory
tens_1 = torch.tensor([0.5, 0.3])
tens_2 = torch.tensor([0.3, 0.2])
tens_3 = torch.tensor([-0.3, -0.5])
tens_4 = torch.tensor([-0.2, 0.5])
# wydruk
print("\n Tensor 1: ", tens_1)
print("\n Tensor 2: ", tens_2)
print("\n Tensor 3: ", tens_3)
print("\n Tensor 4: ", tens_4)

# Obliczamy podobieństwo cosinusowe Cosine Similarity
cosi = torch.nn.CosineSimilarity(dim=0)
output1 = cosi(tens_1, tens_2)
output2 = cosi(tens_1, tens_3)
output3 = cosi(tens_1, tens_4)
# Wynik
print("\n Podobieństwo cosinusowe między Tensorem 1 i Tensorem 2 : ", output1)
print("\n Podobieństwo cosinusowe między Tensorem 1 i Tensorem 3 : ", output2)
print("\n Podobieństwo cosinusowe między Tensorem 1 i Tensorem 4 : ", output3)
```

```python
#Pokażemy, jak ułożone są wektory reprezentowane przez tensory
#zmieniamy tensory na tablice
import matplotlib.pyplot as plt
tens_1_np = tens_1.numpy()
tens_2_np = tens_2.numpy()
tens_3_np = tens_3.numpy()
tens_4_np = tens_4.numpy()
print(tens_1_np, tens_2_np, tens_3_np, tens_4_np)
```

```python
origin = [0, 0]
plt.quiver(*origin, tens_1_np[0], tens_1_np[1], color=['r'], scale=2, label='tens_1')
plt.quiver(*origin, tens_2_np[0], tens_2_np[1], color=['g'], scale=2, label='tens_2')
plt.quiver(*origin, tens_3_np[0], tens_3_np[1], color=['b'], scale=2, label='tens_3')
plt.quiver(*origin, tens_4_np[0], tens_4_np[1], color=['y'], scale=2, label='tens_4')

plt.xlabel('X')
plt.ylabel('Y')
plt.title('Podobieństwo wektorów')
plt.legend()
plt.grid()
plt.show()
```
