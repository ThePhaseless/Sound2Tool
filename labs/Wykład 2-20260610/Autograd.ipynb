## Automatyczne różniczkowanie w PyTorch
torch.autograd

```python
import torch
from torch.autograd import grad
```

$x=(x_1,x_2,x_3,x_4)$

```python
x1=torch.tensor(2.,requires_grad=True,dtype=torch.float32)
x2=torch.tensor(3.,requires_grad=True,dtype=torch.float32)
x3=torch.tensor(1.,requires_grad=True,dtype=torch.float32)
x4=torch.tensor(4.,requires_grad=True, dtype=torch.float32)
```

```python
z1=x1*x2
z2=x3*x4
f=z1+z2
df_dx=grad(outputs=f,inputs=[x1,x2,x3,x4])
```

```python
print(f'gradient względem x1 = {df_dx[0]}')
print(f'gradient względem x2 = {df_dx[1]}')
print(f'gradient względem x3 = {df_dx[2]}')
print(f'gradient względem x4 = {df_dx[3]}')
```
